Data Types 

| Data Type | Meaning                         | Example           |
| --------- | ------------------------------- | ----------------- |
| `int`     | whole number                    | `20`              |
| `float`   | decimal                         | `8.5`             |
| `str`     | text                            | `"Ritu"`          |
| `bool`    | True/False                      | `True`            |
| `list`    | ordered changeable collection   | `[10,20,30]`      |
| `tuple`   | ordered unchangeable collection | `(10,20,30)`      |
| `dict`    | key-value data                  | `{"name":"Ritu"}` |
| `set`     | unique values                   | `{1,2,3}`         |
