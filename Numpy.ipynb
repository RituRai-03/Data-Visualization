9. Mathematical Operations

```python
a=np.array([1,2,3])
b=np.array([4,5,6])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

10. Aggregate Functions

```python
import numpy as np
```

```python
arr = np.array([10,20,30,40])

print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(np.std(arr))
print(np.var(arr))
```

11. Axis Operations

```python
arr=np.array([
    [1,2,3],
    [4,5,6]
])

print(arr.sum(axis=0))
print(arr.sum(axis=1))
```
