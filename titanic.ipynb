```python
data['Sex'] = data["Sex"].replace({
    "male":1,
    "female":0
})
```

```python
import numpy as np
```

```python
data["SurvivalStatus"] = np.where(
    data["Survived"] == 1,
    "Survived",
    "Not Survived"
)
```

```python
data['SurvivalStatus'].value_counts()
```
