```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [100,150,120,180,200]

plt.plot("Months, Sales")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
plt.plot(
    months,
    sales,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o"
)
```

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales_2025 = [100,150,120,180,200]
sales_2026 = [120,170,140,200,220]


plt.plot(months, sales_2025, marker="o")
plt.plot(months, sales_2026, marker="s")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend("center")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Ritu", "Jhalak","Supriya","Rimjhim"]
marks = [85,90,93,95]

plt.bar(students, marks, width=0.6)
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
plt.barh(students, marks)
plt.show()
```

```python
students=["A","B","C","D"]

python =[80,70,90,60]
java =[70,75,85,80]

import numpy as np

x=np.arange(len(students))
width=0.35

plt.bar(x-width/2, python, width, label="Python")
plt.bar(x+width/2, java, width, label="Java")
plt.xticks(x, students)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["AI","Python", "Java", "SQL"]
hours = [35,25,20,20]
explode = [0.1,0,0,0]
mycolors=["Red","hotpink","b","#4CAF50"]

plt.pie(hours,
labels=subjects,
autopct="%1.1f%%",
explode=explode,
shadow=True,
colors=mycolors
)

plt.title("Study Time")
plt.legend()
plt.show()
```

```python
height = [150,155,160,165,170]
weight = [50,55,60,65,70]

plt.scatter(height, weight,
            s=100, #Scatter Plot Size
            alpha=0.5 #Scatter Plot Transparency 
            )

plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

```python
marks=[55,60,65,70,75,80,85,90,95,62,32,79,54]
plt.hist(marks,
bins=5
)
plt.title("Marks Distribution")
plt.show()
```

```python
marks = [55,60,70,65,75,90,98,80,85,81,54,13,78,94,37,58]

plt.boxplot(marks)
plt.title("Marks")
plt.show()
```

```python
python = [70,75,80,85,90]
java = [60,65,70,75,80]
cpp = [65, 70, 75, 80,85]

plt.boxplot(
    [python, java, cpp],
    tick_labels=("Python", "Java", "C++")
)

plt.show()
```
