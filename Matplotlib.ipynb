```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [100,150,120,180,200]

plt.plot("Months, Sales")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
plt.plot(
    months,
    sales,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o"
)
```

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales_2025 = [100,150,120,180,200]
sales_2026 = [120,170,140,200,220]


plt.plot(months, sales_2025, marker="o")
plt.plot(months, sales_2026, marker="s")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend("center")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Ritu", "Jhalak","Supriya","Rimjhim"]
marks = [85,90,93,95]

plt.bar(students, marks, width=0.6)
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
students=["A","B","C","D"]

python =[80,70,90,60]
java =[70,75,85,80]

import numpy as np

x=np.arange(len(students))
width=0.35

plt.bar(x-width/2, python, width, label="Python")
plt.bar(x+width/2, java, width, label="Java")
plt.xticks(x, students)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["AI","Python", "Java", "SQL"]
hours = [35,25,20,20]
explode = [0.1,0,0,0]
mycolors=["Red","hotpink","b","#4CAF50"]

plt.pie(hours,
labels=subjects,
autopct="%1.1f%%",
explode=explode,
shadow=True,
colors=mycolors
)

plt.title("Study Time")
plt.legend()
plt.show()
```

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Titanic-Dataset.csv')
```

```python

data.head(2)
```

```python
avg_fare = data.groupby("Pclass")["Fare"].mean()
print(avg_fare.index)
print(avg_fare.values)
```

```python
avg_fare = data.groupby("Pclass")["Fare"].mean()

plt.plot(avg_fare.index, avg_fare.values, marker="o")

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")

plt.show()
```
